# Interactive Text-to-Image Generation System using Stable Diffusion

## Objective:

To design and implement an interactive AI-based system that generates high-quality images from natural language prompts using the Stable Diffusion diffusion model, allowing users to control image quality, realism, and prompt adherence through configurable parameters

### Importing required libraries

```python
import torch
from diffusers import StableDiffusionPipeline
```

torch is Used for: GPU computation, Data types like float16, Running the model efficiently on CUDA

Stable Diffusion pipeline from Hugging Face Diffusers - This pipeline handles: Text to image generation, Loading model, tokenizer, scheduler, UNet automatically

### Loading the Stable Diffusion model

```python
pipe = StableDiffusionPipeline.from_pretrained("sd-legacy/stable-diffusion-v1-5", torch_dtype= torch.float16).to("cuda")
```

This loop continuously takes user text prompts, generates multiple high-quality images using a Stable Diffusion pipeline with controlled denoising steps and prompt guidance, applies a negative prompt to improve realism, and automatically saves each generated image until the user exits.


```python
while True:
    Prompt = input("Your Prompt: ").strip()

    if Prompt.lower() == "exit":
        print("Session ended")
        break
    Output = pipe(
        Prompt,
        negative_prompt= """low quality, lowres, blurry, out of focus,
                      cartoon, anime, illustration, painting, fantasy,
                      oversaturated colors, unrealistic lighting,
                      harsh shadows, fake shadows, flat lighting,
                      overexposed, underexposed,
                      excessive fog, artificial mist,
                      plastic textures, melted objects,
                      duplicate objects, extra objects,
                      noise, grain, jpeg artifacts,
                      text, watermark, logo, signature""",
                      num_images_per_prompt= 4,
                      num_inference_steps= 50, #Number of denoising steps
                      guidance_scale= 10) #Controls how strongly model follows prompt
    for ind,image in enumerate(Output.images):
      image.save(f"Image_{ind}.png")
```
